```python
import pandas as pd

INPUT_LEN = 2114
HALF_INPUT = INPUT_LEN // 2
```

```python
peak_file = "~/chrombpnet/Hs_TS_immune_chrombpnet/human_immune_peak_raw.modified10col.bed"
chrom_sizes_file = "~/chrombpnet/chrombpnet_tutorial/data/downloads/hg38.chrom.sizes"
output_file = "~/chrombpnet/Hs_TS_immune_chrombpnet/human_immune_peak_valid.modified10col.bed"
```

```python
# chromosome sizes
chrom_sizes = pd.read_csv(
    chrom_sizes_file,
    sep="\t",
    header=None,
    names=["chr", "chrom_size"]
)

chrom_size_dict = dict(
    zip(chrom_sizes["chr"], chrom_sizes["chrom_size"])
)
```

```python
chrom_size_dict
```

```python
# narrowPeak
peaks = pd.read_csv(
    peak_file,
    sep="\t",
    header=None
)

# narrowPeak 第10列是 summit，相对于 start
peaks["center"] = peaks[1] + peaks[9]

peaks["chrom_size"] = peaks[0].map(chrom_size_dict)

valid = (
    peaks["chrom_size"].notna()
    & (peaks["center"] - HALF_INPUT >= 0)
    & (peaks["center"] + HALF_INPUT <= peaks["chrom_size"])
)

print("Total regions:", len(peaks))
print("Valid regions:", valid.sum())
print("Filtered regions:", (~valid).sum())

print("\nFiltered regions:")
print(peaks.loc[~valid, [0, 1, 2, 9, "center", "chrom_size"]])
```

```python
peaks.loc[valid, list(range(10))].to_csv(
    output_file,
    sep="\t",
    header=False,
    index=False
)
```
