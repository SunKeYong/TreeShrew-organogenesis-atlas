```python
%%time
data = snap.pp.import_data(
    path,
    file=[name + '.h5ad' for name in sample],
    chrom_sizes=myCustomGenome,
    sorted_by_barcode=False,
    min_num_fragments=1,
    n_jobs=6,
)
data
```

```python
snap.pp.add_tile_matrix(data, bin_size=500, n_jobs=6)
```

```python
%%time
#[name + '.h5ad' for name in sample]
xdata = snap.AnnDataSet(
    adatas=[(name, adata) for name, adata in zip(sample, data)],
    filename='TS_Downsampl_peak.h5ads'
)
xdata
```

```python
subtype_counts = xdata.obs['sample'].value_counts()

print("sample 的所有取值及细胞数量：")
print(subtype_counts)

# 如果需要看具体有哪些不同的取值
unique_subtypes = xdata.obs['sample'].unique()
print("\n所有不同的 sample：", unique_subtypes)
```

```python
%%time
snap.tl.macs3(xdata, groupby='sample',shift=-75, extsize=150, n_jobs=10, qvalue=0.001)
```

```python
%%time
peaks = snap.tl.merge_peaks(xdata.uns['macs3'], myCustomGenome)
peaks.head()
```

```python
peaks['Peaks'].to_pandas().to_csv('../Peak_files/TS_snapATAC2_macs3_q0.001_peaks.csv')
```

```python
peaks.to_pandas().to_csv('../Peak_files/TS_snapATAC2_macs3_q0.001_peaks_by_celltypes.csv')
```

```python
%%time
snap.ex.export_coverage(xdata, groupby='sample', n_jobs=10, normalization='CPM', out_dir='../BigWig_files')
```

```python
# =======================
# Imports & basic checks
# =======================
import numpy as np
import scanpy as sc
from scipy.sparse import csr_matrix, issparse
from sklearn.feature_extraction.text import TfidfTransformer
from sklearn.decomposition import TruncatedSVD
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics.pairwise import cosine_similarity
import scipy.sparse as sp

# -------------- sanity print --------------
print("Original adata:", adata)   # cells x peaks

# Ensure sparse
if not issparse(adata.X):
    adata.X = csr_matrix(adata.X)
```

```python
# 2) 计算 TF-IDF（cells × peaks）
tfidf = TfidfTransformer(smooth_idf=True, norm='l2')
X_tfidf = tfidf.fit_transform(adata.X)   # 仍是 cells × peaks (稀疏)
print("TF-IDF computed:", X_tfidf.shape)
```

```python
# 3) 转置 → peaks × cells，然后对 peaks 做 TruncatedSVD（LSI），得到 peak embeddings
# transpose: peaks x cells
X_peaks_by_cells = X_tfidf.T.tocsr()   # shape: (n_peaks, n_cells)
print("peaks x cells:", X_peaks_by_cells.shape)
#
# Choose n_components depending on data / downstream (e.g., 50)
n_components = 200
#
# TruncatedSVD on the transposed TF-IDF (this yields peak-level embeddings)
svd = TruncatedSVD(n_components=n_components, random_state=0)
peak_embeddings = svd.fit_transform(X_peaks_by_cells)  # shape: (n_peaks, n_components)
print("peak embeddings shape:", peak_embeddings.shape)
```

```python
# 4) 把 peak embeddings 放回一个新的 AnnData（便于使用 scanpy 的 neighbors/umap 等）
import anndata as ad
# create a peaks-level AnnData: obs corresponds to peaks
peaks_adata = ad.AnnData(X=sp.csr_matrix(X_peaks_by_cells),  # optional, keep raw matrix as X
                         obs=adata.var.copy(),               # peak metadata copied from original adata.var
                         var= None)                          # var unused (no features for peaks)
# add embeddings
peaks_adata.obsm['X_lsi'] = peak_embeddings

print(peaks_adata)
```

```python
sc.pp.neighbors(peaks_adata, use_rep='X_lsi', n_neighbors=50, metric='cosine')
sc.tl.umap(peaks_adata, min_dist=0.2)
```

```python
peaks_adata.obsm['X_umap_50_0.2'] = peaks_adata.obsm['X_umap'].copy()
```

```python
# 提取 'chr1'、'chr2' 等信息
chroms = pd.Series(peaks_adata.obs_names).str.extract(r'^(chr[\w]+)')[0]
```

```python
peaks_adata.obs['chromosome'] = chroms.values

print("✅ Added 'chromosome' to peaks_adata.obs:")
print(peaks_adata.obs['chromosome'].value_counts().head())
```
